```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Define detector layers along the x-axis
n_layers = 10
x_layers = np.linspace(1, 10, n_layers)

# True particle parameters (Initial position y0 = 0.5, True slope = 0.4)
true_y0 = 0.5
true_slope = 0.4

# Generate the "true" clean trajectory of the particle
true_y_trajectory = true_y0 + true_slope * x_layers
# Detector measurement noise (uncertainty sigma)
sigma_noise = 0.15

# Simulate noisy measurements recorded by the detector
np.random.seed(42)  # For reproducible results
measured_y = true_y_trajectory + np.random.normal(
    0, sigma_noise, size=n_layers
)
# State vector: [position y, slope y']
# Initialize starting state estimate
x_state = np.array([measured_y[0], 0.0])

# Covariance matrix (uncertainty of our initial state estimate)
P_cov = np.array([[1.0, 0.0], [0.0, 1.0]])

# Process noise covariance (accounts for small physical perturbations like multiple scattering)
Q = np.array([[0.001, 0.0], [0.0, 0.001]])

# Measurement variance (R)
R = sigma_noise**2

# Measurement matrix H (we only measure position y, not the slope directly)
H = np.array([[1.0, 0.0]])

# Storage for filtered trajectory estimates
filtered_states = []

# Kalman Filter Loop across all detector layers
dx = x_layers[1] - x_layers[0]  #distance between layers

for i in range(n_layers):
  z = measured_y[i]  # Current measurement

  if i > 0:
    # --- 1. PREDICTION STEP ---
    # State Transition Matrix F (y_new = y_old + slope * dx)
    F = np.array([[1.0, dx], [0.0, 1.0]])

    # Predict state
    x_state = F @ x_state
    # Predict covariance
    P_cov = F @ P_cov @ F.T + Q

    # --- 2. UPDATE STEP ---
    # Innovation (residual) between measurement and prediction
    y_residual = z - (H @ x_state)

    # Innovation covariance
    S = H @ P_cov @ H.T + R

    # Kalman Gain K
    K = P_cov @ H.T / S

    # Update state estimate
    x_state = x_state + (K * y_residual).flatten()

    # Update covariance matrix
    I = np.eye(2)
    P_cov = (I - np.outer(K, H)) @ P_cov

  filtered_states.append(x_state[0])  # Save estimated position y
  plt.figure(figsize=(10, 6))
plt.plot(
    x_layers,
    true_y_trajectory,
    "g-",
    linewidth=2,
    label="True Particle Trajectory",
)
plt.scatter(
    x_layers,
    measured_y,
    color="red",
    s=50,
    zorder=3,
    label="Noisy Detector Hits ($z_k$)",
)
plt.plot(
    x_layers,
    filtered_states,
    "b--o",
    linewidth=2,
    label="Kalman Filter Reconstructed Track",
)

plt.xlabel("Detector Layer Position ($x$)", fontsize=12)
plt.ylabel("Coordinate ($y$)", fontsize=12)
plt.title(
    "Toy Kalman Filter Track Reconstruction in Silicon Detector Layers",
    fontsize=14,
)
plt.legend(fontsize=11)
plt.grid(True, linestyle="--", alpha=0.6)

plt.show()
```
